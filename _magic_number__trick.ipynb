```python
import math

# ml2(a, n, k, e): search integers i in [0, n) for one where the fractional
# part of i * a lands in the decimal window [e/10**k, e/10**k + 10**-k).
# e.g. k = 3 and e = 56 means we're looking for a fractional part near 0.056.
# Returns the list of all matching integers i.
def ml2(a, n, k, e):
  a = a - math.floor(a)
  lower_bound = e / 10**k
  upper_bound = lower_bound + 10**(-k)
  u, matches = 0, []
  for i in range(n):
    if lower_bound <= u < upper_bound: matches.append(i)
    u += a
    if u >= 1: u -= 1
  return matches

test = ml2(math.sqrt(2), 10**6, 5, 1001)
hide_value = 77 * math.sqrt(2)
print(hide_value)
print(hide_value * test[0])
# hide_value = 77 * sqrt(2) = 108.89444430272833
# hide_value * 7979 = 868868.7710914693
# the digits right after the decimal point are 77, matching the "hidden" 77 we started with
```
